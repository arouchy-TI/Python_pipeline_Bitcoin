```python
import requests
import pandas as pd
from datetime import datetime

print("Exploracao Iinicial do dados")

# Teste de Extracao
url = "https://api.coinbase.com/v2/prices/spot"
response = requests.get(url)
dados = response.json()
print(f"Resposta Bruta da Api em JSON: {dados}")

# Teste de Tranformacao
preco_usd = float(dados['data']['amount'])
taxa_brl = 5.50
preco_brl = preco_usd * taxa_brl

# Visualizar em dataFrame
df_teste = pd.DataFrame([{
    "timestamo":datetime.now(),
    "moeda":"BTC",
    "usd":preco_usd,
    "brl":preco_brl
}])

print("\n Dados transformados:")
print(df_teste)

```
